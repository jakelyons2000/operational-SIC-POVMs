```python
import math
import numpy
from typing import Tuple
```

```python
def divisors(n: int):
    divs = []
    for i in range(1,int(math.sqrt(n))+1):
        if n % i == 0:
            divs.append(i)
            divs.append(n // i)
    return divs
```

```python
divisors(10)
```

```python
def params(a: int, f: int):
    options = []
    for h in divisors(2*a*f):
        if h > 2:
            for b in divisors((a+h)*f):
                if b < h/2:
                    c = f+2*a*f//h
                    e = (a-b+h)*f//b-c
                    g = h-b
                    options.append((a, b, c, e, f, g))
    return options
```

```python
def boxes(a: int, b: int, c: int, e: int, f: int, g: int):
    boxs = (a*e+a*f+a*g+b*e+b*f+c*e)
    return boxs
```

```python
for a in range(1,100):
    for f in range(1,100):
        if a > 1 | f > 1:
            for tup in params(a,f):
                boxs= boxes(*tup)
                n = math.sqrt(boxs+1)
                rat = good_ratio(*tup)
                if n - rat < 0.1:
                    print(n, rat, tup)
```

```python
def good_ratio(a: int, b: int, c: int, e: int, f: int, g: int):
    ratio1 = a*(a+b+g)*(a+b+c+f+g)/((a+g)*(a+b+f+g)*(a+b+c+e+f+g))
    ratio2 = c*f*(b+f+g)/((c+e)*(b+f)*(a+b+f+g))
    return 1/(ratio1+ratio2)
```

```python
def params2(a: int, b: int):
    options = []
    for c in divisors(b**2):
        e = b**2//c + b
        options.append((a,b,c,e))
    return options
```

```python
def boxes2(a: int, b: int, c: int, e: int):
    boxs = (a**2 + b**2 + 2*a*b + 2*a*c + 2*b*c + 2*a*e)
    return boxs
```

```python
def good_ratio2(a: int, b: int, c: int, e: int):
    ratio = e/(a+e) * b/(b+c) * (b+2*c)/(b+c) * (e+2*c+2*b)/(e+2*b+2*c+a)
    return 1/ratio
```

```python
for a in range(1,1000):
    for b in range(1,1000):
            for tup in params2(a,b):
                boxs= boxes2(*tup)
                n = math.sqrt(boxs + 1)
                rat = good_ratio2(*tup)
                if abs(n - rat) < 0.1:
                    print(n, rat, tup)
```
